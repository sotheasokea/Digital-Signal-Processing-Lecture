```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
vec = [3, 4]

magnitude = np.sqrt(np.sum(np.pow(vec, 2)))
angle = np.rad2deg(np.arctan2(vec[1], vec[0]))

print(magnitude, angle)
```

```python
def time_axis(fs, duration):
    return np.arange(0, duration, 1/fs)

def generate_sine(f, t, a, p):
    return a*np.sin(2*np.pi*f*t + p)
```

```python
# correlation between signal

# we have a signal "y" created with 2 and 3 frequency
t = time_axis(100, 1)
y1= generate_sine(2, t, 1, np.pi)
y2= generate_sine(3, t, 2, 0)

y = y1 + y2

plt.figure(figsize=(15, 5))
plt.subplot(4, 1, 1)
plt.plot(t, y)
plt.legend(['2Hz + 3Hz'])

frequency = [1, 2, 3, 4, 5, 6]
i = 5
for f in frequency:
    y_diff_freq = generate_sine(f, t, 1, 0)
    plt.subplot(4, 4, i)
    i = i + 1
    plt.plot(t, y_diff_freq, label = f"{f}Hz")
    plt.legend()

    r = y*y_diff_freq
    plt.subplot(4, 4, i)
    i = i + 1
    plt.plot(t, r)
    plt.plot(t, np.zeros(len(t)), 'k')
    plt.fill_between(t, r, where = (r>=0), color = "blue", alpha = 0.4)
    plt.fill_between(t, r, where = (r<0), color = "green", alpha = 0.4)
plt.show()
```
