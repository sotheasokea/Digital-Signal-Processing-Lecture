```python
import numpy as np
import matplotlib.pyplot as plt
```

> An equation T{x[n]} is Linear when:
> + T { ax[n] } = a.T{x[n]}
> + T { x1[n] + x2[n] } = T{x1[n]} + T{x2[n]}
>
> or with one condition: T{ ax1[n] + bx2[n] } = a.T{x1[n]} + b.T{x2[n]}

```python
def isLinear(out1, out2):
    if out1 == out2 :
        print("y is linear!")
    else:
        print("y is not linear!")
```

```python
# we have a function for y = 5x
def function_2x(x):
    x = np.array(x)
    return 2*x

a = 2
b = 12
x1 = 2
x2 = 3
out1 = a*function_2x(x1) + b*function_2x(x2)
out2 = function_2x(a*x1 + b*x2)

isLinear(out1, out2)
```

```python
# we have a function for y = x^2
def func_x_toPower2(x):
    x = np.array(x)
    return x**2

a = 2
b = 12
x1 = 2
x2 = 3
out1 = a*func_x_toPower2(x1) + b*func_x_toPower2(x2)
out2 = func_x_toPower2(a*x1 + b*x2)

isLinear(out1, out2)
```

```python
# we have a function for y = 1/n ( x[n] + x[n-1] + x[n-2] )
def func(x):
    x = np.array(x)
    n = len(x)-1
    return 1/n * (x[n] + x[n-1] + x[n-2])

a = 2
b = 12
x1 = [2, 3, 4, 5, 6, 7, 10]
x2 = [10, -4, 5, 8, 1, 2, 2]
out1 = a*func(x1) + b*func(x2)
out2 = func(np.multiply(a,x1) + np.multiply(b, x2))

isLinear(out1, out2)
```
