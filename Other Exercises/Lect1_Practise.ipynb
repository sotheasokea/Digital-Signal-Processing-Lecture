```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
fs= 100
T= 1/fs
duration = 2
A = 1
t = np.arange(0, duration, T)
print(t)
```

```python
f = 1
y = A * np.sin(2*np.pi*f*t)
y1= A * np.sin(2*np.pi*f*t +np.pi/2)
plt.subplot(2, 2, 1)
plt.title('Sin')
plt.xlabel('Time(seconds)')
plt.ylabel('Amplitude')
plt.plot(t, y)
plt.grid()

plt.subplot(2, 2, 2)
plt.title('Cos')
plt.plot(t, y1, color='r')
plt.xlabel('Time(seconds)')
plt.ylabel('Amplitude')
plt.grid()

plt.suptitle("Sinusoid")
plt.tight_layout()
# plt.savefig("myfig.png") --> save the graph as a png image in your current folder
plt.show()
```

```python
# Exercise 6: generate 4 sine waves at difference frequency, amplitude, and phase. plot in 2x2 grid.
# Then merge all 4 waves into one and plot at the third row
# Set the image title to your ID
```

```python
# create a function for generating wave equation
def generate_sine(t, f, A = 1, p = 0):
    y = A * np.sin(2 * np.pi * f * t + p)
    return y

# create function for time vector
def time_axis(fs, start = 0, duration = 1):
    T = 1/fs
    t = np.arange(start, duration, T)
    return t
```

```python
t = time_axis(100, 0, 1)
y1= generate_sine(t, f = 5, A = 4, p = np.pi/2)
y2= generate_sine(t, f = 3, A = 5, p = 0)
y3= generate_sine(t, f = 8, A = 3, p = 3 * np.pi/2)
y4= generate_sine(t, f = 4, A = 2, p = np.pi / 4)

plt.subplot(3, 2, 1)
plt.plot(t, y1, color='r')
plt.title('y1')
plt.xlabel('t')
plt.ylabel('a')
plt.grid()

plt.subplot(3, 2, 2)
plt.plot(t, y2, color='g')
plt.title('y2')
plt.xlabel('t')
plt.ylabel('a')
plt.grid()

plt.subplot(3, 2, 3)
plt.plot(t, y3, color='b')
plt.title('y2')
plt.xlabel('t')
plt.ylabel('a')
plt.grid()

plt.subplot(3, 2, 4)
plt.plot(t, y4, color='c')
plt.title('y4')
plt.xlabel('t')
plt.ylabel('a')
plt.grid()

plt.subplot(3, 1, 3)
plt.plot(t, y1, color='r', label='y1')
plt.plot(t, y2, color='g', label='y2')
plt.plot(t, y3, color='b', label='y3')
plt.plot(t, y4, color='c', label='y4')
plt.title('Merge 4 graph in one plot')
plt.xlabel('Time(seconds)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid()

plt.suptitle("Excercise 6")
plt.tight_layout()
plt.savefig("670710258.png")
plt.show()
```
