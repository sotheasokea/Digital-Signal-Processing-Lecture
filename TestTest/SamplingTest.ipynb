```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.linspace(0, 1, 100)
print(t)

a = np.arange(0, 1, 0.01)
print(a)
```

```python
y = 2 * np.sin(2 * np.pi * 5 * t )
plt.plot(t, y, marker='o', markersize=5, markerfacecolor='r')
plt.grid()
plt.savefig('test.png')
plt.show()
```

```python
t = np.arange(0, 2, 0.01)
y = np.sin(2*np.pi*2*t)
y1 = np.sin(2*np.pi*2*t + np.pi/2)
plt.plot(t, y, marker='o', markersize=2, markerfacecolor='r', label='cos')
plt.plot(t, y1, marker='o', markersize=2, markerfacecolor='b', label='sin')
plt.legend()
plt.grid()
```
