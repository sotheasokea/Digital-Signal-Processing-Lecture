```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
signal = thinkdsp.TriangleSignal(200)
signal.plot()
plt.grid()
```

```python
wave = signal.make_wave(duration = 0.5, framerate = 10000)
spectrum = wave.make_spectrum()
spectrum.plot()
plt.grid()
plt.show()
```

```python
squareSignal = thinkdsp.SquareSignal(100)
squareSignal.plot()
```

```python
wave = squareSignal.make_wave(duration = 0.5, framerate = 10000)
spectrum = wave.make_spectrum()
spectrum.plot()
plt.grid()
plt.show()
```

```python
framerate = 10_000

cos = thinkdsp.CosSignal(4500)
duration = cos.period*5
segement = cos.make_wave(duration, framerate = framerate)
plt.subplot(2, 1, 1)
cos.plot()
segement.plot(label="4500")
plt.legend()

cos = thinkdsp.CosSignal(5500)
segement = cos.make_wave(duration, framerate = framerate)
plt.subplot(2, 1, 2)
cos.plot()
segement.plot(label = "5500")
plt.legend()
```
