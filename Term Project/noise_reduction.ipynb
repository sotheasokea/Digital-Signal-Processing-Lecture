```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from scipy import signal, stft, istft
import IPython.display as ipd
```

```python
fs_clean, clean_signal = wavfile.read("./audio/clean_speech.wav")
t_clean = np.arange(len(clean_signal)) / fs_clean
clean = clean_signal.astype(np.float64)       
clean = clean / np.max(np.abs(clean)) 


fs_noisy, noisy_signal = wavfile.read("./audio/audio_run/noisy_speech.wav")
t_noisy = np.arange(len(noisy_signal)) / fs_noisy
noisy = noisy_signal.astype(np.float64)
noisy = noisy / np.max(np.abs(noisy))
print(fs_clean, fs_noisy, len(clean), len(noisy))
```
