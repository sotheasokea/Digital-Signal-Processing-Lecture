```python
def time_axis(fs , duration):
    T = 1/fs
    t = np.arange(0, duration, T)
    return t
```

```python
t = time_axis(fs = 100, duration = 2)
frequency = 2
phase = np.pi
Amplitude = 2

y = Amplitude * np.sin(2*np.pi*frequency*t + phase)
# lw is the width of ls
plt.plot(t, y, marker ="o", markersize = 10, markerfacecolor = "#11F02F", color = "red", ls = ":", lw = 100)
plt.xlabel("Time(sec)")
plt.ylabel("Amplitude")
plt.grid("True")
plt.show()
```

```python
def generate_sine(t, f, p = 0, a = 1):
    return a*np.sin(2*np.pi*f*t + p)
```

```python
# Exercise 4:
# 1.
t = time_axis(fs = 10000, duration = 1)
f = 1
p = 0
a = 1 
y = generate_sine(t, f, p, a)
plt.plot(t, y, marker = "*", markersize = 20, markerfacecolor="#F54927", color = "black")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid()
plt.show()
```

```python
# 2. 
t = time_axis(fs = 100, duration = 2)
f = 3
p = np.pi/2
a = 2

y = generate_sine(t, f, p, a)
plt.plot(t, y, marker = "D", markersize = 10, markerfacecolor = "#F527E7")
plt.xlabel("Time(sec)")
plt.ylabel("Amplitude")
plt.grid(True)
```

```python
# 3. 
t = time_axis(fs = 250, duration = 1.5)
f = 4
p = 3*np.pi/2
a = 1

y = generate_sine(t, f, p, a)
plt.plot(t, y, marker = "X", markersize = 10, markerfacecolor = "#A12FE7")
plt.xlabel("Time(sec)")
plt.ylabel("Amplitude")
plt.grid(False)
```

```python
# Overlay Plots

t  = time_axis(fs = 1000, duration = 1)
y1 = generate_sine(t, f = 4, p = -np.pi/2, a = 1)
y2 = generate_sine(t, f = 4, p = np.pi/2, a = 1)

plt.plot(t, y1)
plt.plot(t, y2)
plt.grid(True)
```
