Create time axis

```python
import numpy as np
fs = 10
T = 1/fs # 0.1 ==> 0 0.1 0.2 ... 0.9
duration = 5
t = np.arange(0, duration, T)
print(t)
```

```python
fs = 4
T  = 1/fs
duration = 2
t = np.arange(0, duration, T)
print(t) 
```

```python
fs = 100
T = 1/fs
duration = 2
t = np.arange(0, duration, T)
print(t)
```

```python
def time_axis(fs, duration):
    T = 1/fs
    t = np.arange(0, duration, T)
    print("length t = ", len(t))
    return t
```

```python
t = time_axis(fs = 100, duration = 0.5)
print(t)
```

Create sine wave

```python
t = time_axis(fs = 1000, duration = 1)

f = 2
p = 0
y = np.sin(2 * np.pi * f * t + p)

import matplotlib.pyplot as plt
plt.plot(t, y)
plt.xlabel("Time (sec)")
plt.ylabel("Amplitude")
plt.title(f"Sine wave {f} Hz at phase={p}")
plt.grid()
plt.show()
```

```python
plt.plot(t, y, marker='*', markerfacecolor='#3dd6d9',
         markersize=15, color='r', linestyle='-', linewidth = 0.5)
plt.grid()
plt.show()
```

Manipulate multiple plot

```python
def generate_sine(f, t, A=1, p=0):
    y = A*np.sin(2*np.pi*f*t + p)
    return y
```

```python
t = time_axis(fs = 100, duration = 1)
y1 = generate_sine(f=5, t=t, A=1, p=0)
y2 = generate_sine(f=3, t=t, A=1, p=0)
```

```python
# Plot in single image
t = time_axis(fs = 100, duration = 1)
y1 = generate_sine(f=5, t=t, A=1, p=0)
y2 = generate_sine(f=5, t=t, A=2, p=np.pi)
plt.plot(t, y1, color='b')
plt.plot(t, y2, color='r')
plt.legend( ['y1', 'y2'] )
plt.title("2 sine wave")
plt.xlabel("time (sec)")
plt.ylabel("amplitude")
plt.grid()
plt.show()
```

```python
# With subplot
plt.subplot(1,2,1) # nRow, nCol, index
plt.plot(t, y1, color='b')
plt.title("y1 wave")
plt.grid()

plt.subplot(1,2,2)
plt.plot(t, y2, color='r')
plt.ylabel("Amplitude")
plt.xlabel("time (sec)")
plt.grid()

plt.suptitle("Sine wave")
plt.tight_layout()
plt.show()
```

Wave inteference

```python
t = time_axis(fs = 100, duration = 1)
y1 = generate_sine(f=5, t=t, A=1, p=0)
y2 = generate_sine(f=3, t=t, A=2, p=0)

plt.subplot(3,1,1)
plt.plot(t, y1, color='b')
plt.title("y1 wave")
plt.grid()

plt.subplot(3,1,2)
plt.plot(t, y2, color='r')
plt.title("y2 wave")
plt.grid()

y = y1+y2
plt.subplot(3,1,3)
plt.plot(t, y, color='k')
plt.title("y wave")
plt.grid()

plt.tight_layout()
plt.show()
plt.savefig("myfig.png")
```
